```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_scores.csv")
print(df.head())

ones = np.ones(len(df))
X = np.column_stack([ones, df['Hours']])
print(X)

theta = np.linalg.inv(X.T @ X) @ (X.T @ df['Scores'])
print(theta)

check = np.polyfit(df['Hours'], df['Scores'], 1)
print(check)

x = df['Hours']
y = df['Scores']
y_hat = X @ theta
print(y_hat)

plt.scatter(x, y, s=80, label="Actual data")
plt.plot(x, y_hat, color="red", linewidth=2)
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Normal Eqn Line")
plt.legend()
plt.show()

y_cap = theta[0] + 7.5 * theta[1]
print(y_cap)
```
