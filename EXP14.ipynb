```python
import numpy as np
import pandas as pd
import time
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
```

```python
train=pd.read_csv("fashion-mnist_train.csv")
test=pd.read_csv("fashion-mnist_test.csv")
```

```python
config=[
    {"learning_rate_init":0.001,"batch_size":32,"max_iter":20},
    {"learning_rate_init":0.01,"batch_size":32,"max_iter":20},
    {"learning_rate_init":0.001,"batch_size":128,"max_iter":20},
    {"learning_rate_init":0.001,"batch_size":32,"max_iter":50}
]
rows=[]
```

```python
X_train=(train.drop(["label"],axis=1))/255.0
y_train=train["label"]

X_test=(test.drop(["label"],axis=1))/255.0
y_test=test["label"]
```

```python
for c in config:
    mpl=MLPClassifier(
        hidden_layer_sizes=(100,),
        activation="relu",
        solver="adam",
        random_state=42,
        **c
    )
    start=time.time()
    mpl.fit(X_train,y_train)
    elapsed_time=time.time()-start
    y_pred=mpl.predict(X_test)
    accuracy=accuracy_score(y_pred,y_test)
    rows.append({
        **c,
        "Accuracy":accuracy*100,
        "Iterations Used":mpl.n_iter_,
        "Final Loss":mpl.loss_
    })
```
