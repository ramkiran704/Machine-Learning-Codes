```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB, BernoulliNB

from sklearn.metrics import accuracy_score, f1_score
from sklearn.metrics import classification_report
```

```python
train = pd.read_csv("train_20newsgroups.csv")
test = pd.read_csv("test_20newsgroups.csv")

print("Training Data:")
print(train.head())

print("\nTesting Data:")
print(test.head())

print("\nTraining Shape:", train.shape)
print("Testing Shape:", test.shape)
```

```python
X_train = train["documents"]
y_train = train["labels"]

X_test = test["documents"]
y_test = test["labels"]

print("Training documents:", len(X_train))
print("Testing documents:", len(X_test))

print("Number of categories:", y_train.nunique())
```

```python
# Convert text into word frequency vectors
vectorizer1 = CountVectorizer(
    stop_words="english",
    max_features=100000
)

X_train_multi = vectorizer1.fit_transform(X_train)
X_test_multi = vectorizer1.transform(X_test)

# Create Multinomial Naive Bayes model
mnb = MultinomialNB()

# Train the model
mnb.fit(X_train_multi, y_train)

# Predict the output
y_pred_multi = mnb.predict(X_test_multi)

# Calculate performance
multi_accuracy = accuracy_score(y_test, y_pred_multi)

multi_f1 = f1_score(
    y_test,
    y_pred_multi,
    average="weighted"
)

print("Multinomial Naive Bayes")
print("Accuracy:", multi_accuracy)
print("F1 Score:", multi_f1)
```

```python
# Convert text into binary word-presence vectors
vectorizer2 = CountVectorizer(
    stop_words="english",
    binary=True,
    max_features=100000
)

X_train_bern = vectorizer2.fit_transform(X_train)
X_test_bern = vectorizer2.transform(X_test)

# Create Bernoulli Naive Bayes model
bnb = BernoulliNB()

# Train the model
bnb.fit(X_train_bern, y_train)

# Predict the output
y_pred_bern = bnb.predict(X_test_bern)

# Calculate performance
bern_accuracy = accuracy_score(y_test, y_pred_bern)

bern_f1 = f1_score(
    y_test,
    y_pred_bern,
    average="weighted"
)

print("Bernoulli Naive Bayes")
print("Accuracy:", bern_accuracy)
print("F1 Score:", bern_f1)
```

```python
results = pd.DataFrame({
    "Model": [
        "Multinomial Naive Bayes",
        "Bernoulli Naive Bayes"
    ],

    "Accuracy": [
        multi_accuracy,
        bern_accuracy
    ],

    "F1 Score": [
        multi_f1,
        bern_f1
    ]
})

print(results)
```
