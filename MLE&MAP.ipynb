```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
data =load_breast_cancer()
```

```python
x=data.data
y=data.target
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
scalar=StandardScaler()
x_train=scalar.fit_transform(x_train)
x_test=scalar.transform(x_test)
```

```python
mle=LogisticRegression(penalty=None,max_iter=5000)
mle.fit(x_train,y_train)
```

```python
l1=LogisticRegression(penalty='l1',solver='liblinear',C=1)
l1.fit(x_train,y_train)
```

```python
l2=LogisticRegression(penalty='l2',C=1,max_iter=5000)
l2.fit(x_train,y_train)
```

```python
pred_mle=mle.predict(x_test)
pred_l1=l1.predict(x_test)
pred_l2=l2.predict(x_test)
```

```python
print("Accuracy of MLE:",accuracy_score(y_test,pred_mle))
print("Accuracy of L1:",accuracy_score(y_test,pred_l1))
print("Accuracy of L2:",accuracy_score(y_test,pred_l2))
```

```python
print("MLE:",mle.coef_[0][:5])
print("L1:",l1.coef_[0][:5])
print("L2:",l2.coef_[0][:5])
```

```python
print("MLE:",np.sum(mle.coef_)!=0)
print("L1:",np.count_nonzero(l1.coef_))
print("L2:",np.count_nonzero(l2.coef_))
```
